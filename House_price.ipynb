```python
import pandas as pd
import numpy as np
```

```python
df= pd.read_csv('final_dataset.csv')
```

```python
df
```

```python
df.shape
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

x_scaler = StandardScaler()
y_scaler = StandardScaler()

df = df.drop(columns=['Unnamed: 0'])

x = df.drop(columns=['price'])
y = df['price']

x_train, x_test, y_train, y_test= train_test_split(x,y, test_size=0.2)
```

```python
x_train_scaled= x_scaler.fit_transform(x_train)
x_test_scaled= x_scaler.transform(x_test)

y_train_scaled = y_scaler.fit_transform(y_train.to_numpy().reshape(-1, 1))
y_test_scaled = y_scaler.transform(y_test.to_numpy().reshape(-1, 1))
```

```python
x_train_scaled
```

```python
x_train_scaled.shape
```

```python
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout

model= Sequential()

model.add(Input((4,)))
model.add(Dense(128, activation='relu'))
model.add(Dropout(0.1))
model.add(Dense(128, activation='relu'))
model.add(Dense(1))
```

```python
from tensorflow.keras.optimizers import Adam

optimizer = Adam(learning_rate=0.001)
model.compile(optimizer= optimizer, loss='mse', metrics=['mae','r2_score'])
```

```python
model.fit(x_train_scaled,y_train_scaled, epochs=30)
```

```python
print(y_train.min())
print(y_train.max())
print(y_train.mean())
print(y_train.std())

model.summary()
```

```python
model.evaluate(x_test_scaled,y_test_scaled)
```

```python
y_pred_scaled = model.predict(x_test_scaled)

y_pred = y_scaler.inverse_transform(y_pred_scaled)

from sklearn.metrics import mean_absolute_error, r2_score

print("MAE:", mean_absolute_error(y_test, y_pred))
print("R²:", r2_score(y_test, y_pred))
```
